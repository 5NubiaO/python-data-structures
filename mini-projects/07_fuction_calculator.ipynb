```python
from logging import exception
#Rebuilt the calculator with function implementation

def add(a,b):
  result = a+b
  return result

def substract(a,b):
  result = a-b
  return result

def mult(a,b):
  result = a*b
  return result

def div(a,b):
  try:
    result = a/b
    return result
  except ZeroDivisionError:
    print('Cannot divide by Zero')


while True:

    print('Choose an operation: +, -, *, / or finish to exit:')
    op = input()
    if op == 'finish':
      print('Bye ^^')
      break
    if op not in ['+','-','*','/']:
      print('Not valid Operation. Try again')
      continue


    try:
      a = (float(input('First Number: ')))
      b = (float(input('Second Number: ')))
    except:
      print('Invalid input')
      continue

    if op == '+':
      result = add(a,b)
      print('The result is: '+ str(result))
      continue

    elif op == '-':
      result = substract(a,b)
      print('The result is: '+ str(result))
      continue

    elif op == '*':
      result = mult(a,b)
      print('The result is: '+ str(result))
      continue

    elif op == '/':
      result = div(a,b)
      if b == 0:
        print('Sorry, try again')
        continue
      else:
        print('The result is: '+ str(result))
        continue
```
