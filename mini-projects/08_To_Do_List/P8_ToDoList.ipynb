```python

#This programm is a to do list

#Ask the user what to do and exit when 4

import sys
Tasks =[]

def Add():
  print('Whats your new task?')
  newT = input()
  Tasks.append(newT)
  return Tasks[-1]

def View(items):
  if not items:
    return 'Nothing yet'
  if len(items)==1:
    return str('[]'+items[0])

  show ="\n".join(str(item) for item in items [:])
  return show

def Delete():
  if not Tasks:
    print('Nothing to delete yet')
  else:
    print('Wich one do you want to delete?')
    for index, value in enumerate(Tasks):
      print(f"[{index +1}]  {value}")

    try:
      er = int(input())
      er -= 1
      posicion = Tasks[er]
      Tasks.remove(posicion)
      print(f"Task {posicion} deleted" )
    except (ValueError, IndexError):
      print('Such Task doesnt exist, Try again...')


while True:
  print('Hi, What do you want to do?')
  print('1. Add a Task')
  print('2. View Tasks')
  print('3. Delete Task')
  print('4. Exit')
  try:
    op = int(input())
  except ValueError:
    print('Invalid Option, try again...')
    continue

  match op:
    case 1:
      print('Add')
      Add()
      print('Your new Task is')
      print(Tasks[-1])

    case 2:
      print('Your Tasks:')
      show = View(Tasks)
      print(show)
    case 3:
      Delete()
    case 4:
      sys.exit()

```
