```python
#This programs ask the user how many days they have lived based on their age

#Basic Version

age=(int(input('How old are you?:' )))
livedyears = age * 365
print('OMG you have lived '+ str(int(livedyears))+ ' days!, Feel Happy about it')

```

```python
#Advanced Version

import calendar
age=(int(input('How old are you?:' )))

actualYear  = calendar.datetime.datetime.now().year #This one tells the computer on what year are we
born = actualYear - age #This one tells the computer where to start from
dayslived=0


for year in range(born,actualYear):
  if calendar.isleap(year):
   dayslived += 366
  else:
    dayslived += 365

print('OMG you have lived '+str(int(dayslived))+' days')

```
